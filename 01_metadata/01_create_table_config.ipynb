```python
CATALOG = "ccap_md"

METADATA_SCHEMA = f"{CATALOG}.metadata"

print(f"Metadata schema: {METADATA_SCHEMA}")
```

```python
from pyspark.sql.types import (
    StructType,
    StructField,
    StringType,
    BooleanType
)
table_config_schema=StructType(
    [
        StructField("table_name",StringType(),False),
        StructField("source_name",StringType(),False),
        StructField("target_name",StringType(),False),
        StructField("load_type",StringType(),False),
        StructField("incremental",BooleanType(),False)
    ]
)
table_config_data=[
    ("customers","customers","customers","FULL",True),
    ("accounts","accounts","accounts","INCREMENTAL",True),
    ("transactions","transactions","transactions","INCREMENTAL",True)
]
table_config_df = spark.createDataFrame(table_config_data,table_config_schema)
display(table_config_df)
```

```python
TABLE_NAME = f"{METADATA_SCHEMA}.table_config"

table_config_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(TABLE_NAME)

print(f"Created: {TABLE_NAME}")
```

```python
table_config = spark.table(
    f"{METADATA_SCHEMA}.table_config"
)

display(table_config)
```
