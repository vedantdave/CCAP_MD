```python
CATALOG = "ccap_md"

METADATA_SCHEMA = f"{CATALOG}.metadata"

print(f"Metadata schema: {METADATA_SCHEMA}")
```

```python
from pyspark.sql.types import (
    StructType,
    StructField,
    StringType,
    BooleanType
)

column_config_data = [

    # --------------------------------------------------------
    # CUSTOMERS
    # --------------------------------------------------------

    ("customers", "customer_id", "STRING", False, True),
    ("customers", "first_name", "STRING", False, True),
    ("customers", "last_name", "STRING", False, True),
    ("customers", "email", "STRING", True, True),
    ("customers", "date_of_birth", "DATE", True, True),
    ("customers", "status", "STRING", False, True),

    # --------------------------------------------------------
    # ACCOUNTS
    # --------------------------------------------------------

    ("accounts", "account_id", "STRING", False, True),
    ("accounts", "customer_id", "STRING", False, True),
    ("accounts", "account_type", "STRING", False, True),
    ("accounts", "account_status", "STRING", False, True),
    ("accounts", "credit_limit", "DECIMAL(18,2)", False, True),
    ("accounts", "opened_date", "DATE", False, True),

    # --------------------------------------------------------
    # TRANSACTIONS
    # --------------------------------------------------------

    ("transactions", "transaction_id", "STRING", False, True),
    ("transactions", "customer_id", "STRING", False, True),
    ("transactions", "account_id", "STRING", False, True),
    ("transactions", "transaction_ts", "TIMESTAMP", False, True),
    ("transactions", "amount", "DECIMAL(18,2)", False, True),
    ("transactions", "currency", "STRING", False, True),
    ("transactions", "transaction_type", "STRING", False, True),
    ("transactions", "status", "STRING", False, True)
]

column_config_schema = StructType([
    StructField("table_name", StringType(), False),
    StructField("column_name", StringType(), False),
    StructField("data_type", StringType(), False),
    StructField("nullable", BooleanType(), False),
    StructField("active", BooleanType(), False)
])
```

```python
column_config_df = spark.createDataFrame(
    column_config_data,
    column_config_schema
)

display(column_config_df)
```

```python
TABLE_NAME = f"{METADATA_SCHEMA}.column_config"

column_config_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(TABLE_NAME)

print(f"Created: {TABLE_NAME}")
```

```python
column_config = spark.table(
    f"{METADATA_SCHEMA}.column_config"
)

display(column_config)
```

```python
display(
    column_config
    .groupBy("table_name")
    .count()
    .orderBy("table_name")
)
```
